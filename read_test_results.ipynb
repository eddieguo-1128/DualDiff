```python
import numpy as np
import os
import glob
import pandas as pd
import warnings


warnings.filterwarnings("ignore", category=FutureWarning)


def print_test_results(mixing_layer=str, warm_up_epoch=str, random_ratio=str, seed=str):
    
    log_dir = os.path.join(
        "framework",
        f"mixup_ablation_mixing_layer_{mixing_layer}_warm_up_epoch_{warm_up_epoch}_random_ratio_{random_ratio}_s{seed}_zoption2",
        "logs"
    )

    result_files = glob.glob(os.path.join(log_dir, "test_results_*.npy"))
    
    if not result_files:
        print(f"[!] No result file found in {log_dir}\n")
        return
    
    file_path = result_files[0]  # Use the first matching file
    
    data = np.load(file_path, allow_pickle=True).item()
    df = pd.DataFrame(data).T

    if "model_path" in df.columns:
        df = df.drop(columns=["model_path"])

    df = df.applymap(lambda x: round(float(x), 4) if isinstance(x, (np.floating, float)) else x)

    print(f"Mixing Layer: {mixing_layer}, Warm-up Epoch: {warm_up_epoch}, Random Ratio: {random_ratio}, Seed: {seed}")
    print(df)
    print("\n")

    return


mixing_layers  = ["-1", "0", "1", "2", "3", "4"]
warm_up_epochs = ["100", "150"]
random_ratios  = ["False", "True"]
seeds          = ["42", "43", "44"]

for mixing_layer in mixing_layers:
    for warm_up_epoch in warm_up_epochs:
        for random_ratio in random_ratios:
            for seed in seeds:
                print_test_results(mixing_layer, warm_up_epoch, random_ratio, seed)
```
